```python
from meshnet.planner.optimizer import HeuristicBasedGraspOptimizer
from meshnet.utils.gripper import RobotiqHandEGripper
from meshnet.utils.mesh import Mesh
from meshnet.utils.plotter import plot_grasp

mesh = Mesh.read("../resources/meshes/train/msh/L-Bracket1.msh")

gripper = RobotiqHandEGripper()
optimizer = HeuristicBasedGraspOptimizer(gripper)
grasp = optimizer.optimize(mesh, mu=0.5, n_samples=100)
plot_grasp(mesh, grasp, gripper)
```

```python
from meshnet.planner.sampler import GraspSampler
from meshnet.utils.plotter import plot_grasp

sampler = GraspSampler(mesh.surface, gripper, mu=1.0)
candidates = sampler.sample(n_samples=100)
plot_grasp(mesh, candidates[0], gripper)
```
